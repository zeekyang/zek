```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from astropy import units as u
from astropy import constants as const
```

```python
df = pd.read_csv("../../course_assignments/homework9/solar_system.csv")
df_original = df.copy()
df_original
```

```python
df = df.set_index("Attribute").T
df.index.name = "Planet"
df.reset_index(inplace=True)
df.columns.name = None
df
```

```python
# 2.1.3
exclude = ["Planet", "Ring System?", "Global Magnetic Field?"]

cols_to_convert = [col for col in df.columns if col not in exclude]

for col in cols_to_convert:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df.dtypes

# a)
# float64, int64 and object

# b)
# float64: 64-bit floating-point number
# int64: 64-bit integer
# object: Generic object type, cannot be expressed as number

# c)
# "Planet", "Ring System?" and "Global Magnetic Field?" are still objects since they contain text labels and not numbers
```

```python
def attach_units(col_name, unit, new_col_name=None):
        df[col_name] = df[col_name].apply(lambda x: x * unit)
    
        if new_col_name is not None:
            df.rename(columns={col_name: new_col_name}, inplace=True)

attach_units(
    col_name="Mass (10^24kg)",
    unit=(1e24 * u.kg),
    new_col_name="Mass (kg)"
)

print(df["Mass (kg)"])
```

```python


# Compute semi-major axis = average of perihelion + aphelion
semi_major_axis = (df["Perihelion (10^6 km)"] + df["Aphelion (10^6 km)"]) / 2

# Attach units: values are in (10^6 km)
semi_major_axis = semi_major_axis * 1e6 * u.km

# Insert new column RIGHT AFTER Aphelion (which is index 9)
df.insert(
    loc=df.columns.get_loc("Aphelion (10^6 km)") + 1,
    column="Semi-Major Axis (km)",
    value=semi_major_axis
)

# Print all column names to confirm insertion
print("\nCOLUMN NAMES AFTER INSERTING SEMI-MAJOR AXIS:")
print(df.columns)

# Print the new column
print("\nSemi-Major Axis Column (with units):")
print(df["Semi-Major Axis (km)"])


# -----------------------------
# 2.3.2  ORBITAL PERIOD → YEARS
# -----------------------------

# Convert from days to years using Astropy

orbital_period_years = (df["Orbital Period (days)"].values * u.day).to(u.year)

# Assign back to DataFrame
df["Orbital Period (yr)"] = orbital_period_years

# Rename the column
df.rename(columns={"Orbital Period (days)": "Orbital Period (yr)"}, inplace=True)

# Print column names
print("\nCOLUMN NAMES AFTER RENAMING ORBITAL PERIOD:")
print(df.columns)

# Print the full column
print("\nOrbital Period (yr):")
print(df["Orbital Period (yr)"])

# Print your favorite planet’s orbital period
favorite = "Earth"    # ← CHANGE IF YOU WANT A DIFFERENT PLANET
value = df.loc[df["Planet"] == favorite, "Orbital Period (yr)"].iloc[0]
print(f"\n{favorite}'s orbital period in years: {value:.4f}")


# -----------------------------
# 2.3.3  DISTANCE (km → AU)
# -----------------------------

# Define: What is 1 AU?
print("\nDEFINITION:")
print("One Astronomical Unit (AU) is the average distance from the Earth to the Sun.")

# Print 1 AU (astropy)
print("\n1 AU with Astropy:")
print(const.au)

# How many kilometers is 1 AU?
print("\nKilometers in 1 AU:")
print(const.au.to(u.km))

# Columns to convert (all are in km or 10^6 km)
distance_cols = [
    "Diameter (km)",
    "Distance from Sun (10^6 km)",
    "Perihelion (10^6 km)",
    "Aphelion (10^6 km)",
    "Semi-Major Axis (km)"
]

# Convert each column to AU
for col in distance_cols:

    # Get numerical values (some columns are in 10^6 km, so scale them)
    if "10^6" in col:
        values_km = df[col] * 1e6 * u.km
    else:
        values_km = df[col] * u.km

    # Convert to AU
    values_au = values_km.to(u.AU)

    # Build new name
    new_name = col.replace("(km)", "(AU)").replace("(10^6 km)", "(AU)")

    # Store back in DataFrame
    df[col] = values_au
    df.rename(columns={col: new_name}, inplace=True)

# Print sample values for your favorite planet
print(f"\nDistance column samples for {favorite}:")
for col in df.columns:
    if "(AU)" in col:
        val = df.loc[df["Planet"] == favorite, col].iloc[0]
        print(f"{col}: {val}")
```
