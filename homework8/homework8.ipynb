```python
median_lat = df['latitude'].median()
median_long = df['longitude'].median()
df['distance_to_median_lat'] = abs(df['latitude'] - median_lat)
df['distance_to_median_long'] = abs(df['longitude'] - median_long)
df.head()
```

```python
df.shape
```

```python
closest_lat = df[df['distance_to_median_lat'] == df['distance_to_median_lat'].min()]
farthest_lat = df[df['distance_to_median_lat'] == df['distance_to_median_lat'].max()]

closest_long = df[df['distance_to_median_long'] == df['distance_to_median_long'].min()]
farthest_long = df[df['distance_to_median_long'] == df['distance_to_median_long'].max()]
print(f"State closest to median latitude: {closest_lat['state'].values[0]} ({closest_lat['distance_to_median_lat'].values[0]:.2f})")
print(f"State farthest from median latitude: {farthest_lat['state'].values[0]} ({farthest_lat['distance_to_median_lat'].values[0]:.2f})")

print(f"State closest to median longitude: {closest_long['state'].values[0]} ({closest_long['distance_to_median_long'].values[0]:.2f})")
print(f"State farthest from median longitude: {farthest_long['state'].values[0]} ({farthest_long['distance_to_median_long'].values[0]:.2f})")
```

```python
# From the Pandas documentation site , I click on “API reference” and “General functions” and find a function we haven’t used yet in class: pd.cut().
# The pd.cut() function divides continuous numeric data into bins or intervals, allowing you to convert numeric variables into categorical ones.

df['latitude_category'] = pd.cut(df['latitude'], bins=3, labels=['Low', 'Medium', 'High'])

print(df[['state', 'latitude', 'latitude_category']].head())
```
