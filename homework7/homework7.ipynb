```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 2.1
def make_sine_wave(x, A, w):
    y = A * np.sin(w * x)
    return y
    
# 2.2
x = np.linspace(0, 2 * np.pi, 1000)

# 2.3
amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]

# 2.4
fig, axs = plt.subplots(1, 3, figsize=(25, 5))

# 2.5 + 2.6
for A, w in zip(amplitude, frequency):
    y1 = make_sine_wave(x, A, 1)
    axs[0].plot(x, y1, label=f"A={A}, ω=1")

    y2 = make_sine_wave(x, 1, w)
    axs[1].plot(x, y2, label=f"A=1, ω={w}")

    y3 = make_sine_wave(x, A, w)
    axs[2].plot(x, y3, label=f"A={A}, ω={w}")

# 2.7
axs[0].set_title("Varying Amplitude (ω=1)", fontsize=16)
axs[1].set_title("Varying Frequency (A=1)", fontsize=16)
axs[2].set_title("Varying Amplitude and Frequency Together", fontsize=16)

for ax in axs:
    ax.set_xlabel("x", fontsize=14)
    ax.set_ylabel("y", fontsize=14)
    ax.legend(fontsize=12)
    ax.grid(True)

#  2.8
plt.show()

```
