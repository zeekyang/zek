```python
import numpy as np
```

```python
# 3.1

list = [20, 27, 9, 12, 31, 17, 4, 50, 39, 35]

arr = np.array(list)
arr_x4 = arr * 4
print(arr_x4)

mean_val = np.mean(arr)
print(mean_val)

median_val = np.median(arr)
print(median_val)

std_val = np.std(arr)
print(std_val)

min_index = np.argmin(arr)
max_index = np.argmax(arr)
print(min_index)
print(max_index)

print(arr[min_index])
print(arr[max_index])

min_val = np.min(arr)
max_val = np.max(arr)
print(min_val)
print(max_val)
# they match

greater_than_20 = arr[arr > 20]
print(greater_than_20)
```

```python
# 3.2

matrix = np.array([
    [5, 12, 17, 9],
    [8, 21, 14, 3],
    [6, 13, 19, 11]
])

print(matrix.shape)
print(matrix.shape[0])
print(matrix.shape[1])

print(matrix[:, 2])

print(matrix[1])

matrix_x2 = matrix * 2
print(matrix_x2)

col_means = np.mean(matrix, axis=0)
print(col_means)

overall_mean = np.mean(matrix)
print(overall_mean)

row_max = np.max(matrix, axis=1)
print(row_max)

col_min = np.min(matrix, axis=0)
print(col_min)

flat = matrix.flatten()
print(flat)
print(flat.shape)
# It flattened the 2D array into 1D containing all elements in their row order.

min_index_flat = np.argmin(flat)
max_index_flat = np.argmax(flat)
print(min_index_flat)
print(max_index_flat)

min_coords = np.unravel_index(min_index_flat, matrix.shape)
max_coords = np.unravel_index(max_index_flat, matrix.shape)
print(min_coords)
print(max_coords)

greater_than_15 = matrix[matrix > 15]
print(greater_than_15)

sorted_matrix = np.sort(matrix, axis=1)
print(sorted_matrix)
```

```python
# 4

arr = np.array([
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13]
])

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(np.sqrt(n)) + 1):
        if n % i == 0:
            return False
    return True

rows_with_prime = []
for row in arr:
    if any(is_prime(x) for x in row):
        rows_with_prime.append(row)

rows_with_prime = np.array(rows_with_prime)

print(arr)
print(rows_with_prime)
```

```python
# 5.2 & 5.3

def checkerboard():
    board = np.zeros((8, 8))
    board[::2, ::2] = 1
    board[1::2, 1::2] = 1
    return board

print(checkerboard())
```

```python
# 5.4

def reverse_checkerboard():
    board = np.zeros((8, 8), dtype=int)
    board[::2, 1::2] = 1  
    board[1::2, ::2] = 1
    return board

print(reverse_checkerboard())
```

```python
# 6

def expansion(arr, n):
    space = " " * n
    expanded = [space.join(word) for word in arr]
    return np.array(expanded)

universe = np.array(['galaxy', 'clusters'])

print(expansion(universe, 1))
```

```python
# 7

def second_brightest(stars):
    sorted_rows = np.sort(stars, axis=1)
    return sorted_rows[:, -2]

np.random.seed(123)
stars = np.random.randint(500, 2000, (5, 5))
print(stars)
print(second_brightest(stars))
```
